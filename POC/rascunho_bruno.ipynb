```python
# A função str.extract() usa a expressão regular (regex) para extrair os nomes dos convênios.
# O padrão r'autorizacao-([a-zA-Z_]+)-' captura o texto entre 'autorizacao-' e o hífen seguinte.
# O 'na=False' garante que a extração não seja aplicada em linhas onde o link é nulo (NaN).
df['CONVENIO'] = df['DS_GUIA_PATH'].str.extract(r'autorizacao-([a-zA-Z_]+)-', expand=False)

# Exibe as 5 primeiras linhas da nova coluna para conferência
print("--- Amostra da nova coluna 'CONVENIO' ---")
display(df[['DS_GUIA_PATH', 'CONVENIO']].head())

# Exibe as 5 últimas linhas também, para vermos a variedade
print("\n--- Últimas 5 linhas para comparação ---")
display(df[['DS_GUIA_PATH', 'CONVENIO']].tail())
```

```python
# Conta a ocorrência de cada convênio na nova coluna.
contagem_convenios = df['CONVENIO'].value_counts()

# Conta o número de links ausentes (NaN) na coluna original.
contagem_na = df['DS_GUIA_PATH'].isna().sum()

print("--- Contagem por Convênio (Top 40) ---")
display(contagem_convenios.head(40))

print(f"\nNúmero total de links ausentes (NA): {contagem_na}")
```

```python
# Pega a contagem de convênios que já tínhamos e transforma em um DataFrame
df_contagem = contagem_convenios.reset_index()
df_contagem.columns = ['CONVENIO', 'FREQUENCIA']

# Calcula o percentual de cada convênio em relação ao total de links extraídos
total_extraido = df_contagem['FREQUENCIA'].sum()
df_contagem['PERCENTUAL (%)'] = round((df_contagem['FREQUENCIA'] / total_extraido) * 100, 2)

# Pega a nossa tabela de exemplos de links
df_exemplos = df.dropna(subset=['CONVENIO']).groupby('CONVENIO')['DS_GUIA_PATH'].first().reset_index()
df_exemplos.rename(columns={'DS_GUIA_PATH': 'LINK_EXEMPLO'}, inplace=True)

# Agora, une as duas tabelas usando a coluna 'CONVENIO' como chave
df_final_detalhado = pd.merge(df_contagem, df_exemplos, on='CONVENIO')

# Garante que a coluna de links seja exibida completamente
pd.set_option('display.max_colwidth', None)


# Função para converter um texto de URL em uma tag HTML de link clicável
def tornar_clicavel(url):
    # Retorna a tag <a> do HTML. target="_blank" faz com que o link abra em uma nova aba.
    return f'<a href="{url}" target="_blank">{url}</a>'

# Prepara a tabela para exibição com o estilo
# Usamos .format() para aplicar a função 'tornar_clicavel' na coluna 'LINK_EXEMPLO'
# e também podemos formatar a coluna de percentual para exibir o símbolo '%'
tabela_formatada = df_final_detalhado.style.format({
    'LINK_EXEMPLO': tornar_clicavel,
    'PERCENTUAL (%)': '{:.2f}%'.format
})

print("--- Tabela Detalhada com Links Clicáveis ---")

# Exibe a tabela estilizada
display(tabela_formatada)
```

```python
# Filtra o DataFrame original para manter apenas as linhas onde:
# 1. 'CD_AVISO' não é Nulo/NA (df['CD_AVISO'].notna())
# 2. 'CD_AVISO' é maior que zero (df['CD_AVISO'] > 0)
df_com_aviso_valido = df[(df['CD_AVISO'].notna()) & (df['CD_AVISO'] > 0)]

# Agora, fazemos a contagem de convênios nesse DataFrame filtrado
contagem_convenio_com_aviso = df_com_aviso_valido['CONVENIO'].value_counts()

# Exibe o resultado
print("--- Contagem de Convênios (Apenas com CD_AVISO > 0) ---")
print(f"Total de guias com CD_AVISO válido: {len(df_com_aviso_valido)}")
print("-" * 50)

# 1. Pega a contagem da célula anterior e converte para um DataFrame
df_contagem_aviso = contagem_convenio_com_aviso.reset_index()
df_contagem_aviso.columns = ['CONVENIO', 'FREQUENCIA']

# 2. Calcula o percentual em relação ao novo total (apenas guias com aviso válido)
total_com_aviso = df_contagem_aviso['FREQUENCIA'].sum()
df_contagem_aviso['PERCENTUAL (%)'] = round((df_contagem_aviso['FREQUENCIA'] / total_com_aviso) * 100, 2)

# 3. Pega um link de exemplo de dentro do DataFrame já filtrado
df_exemplos_aviso = df_com_aviso_valido.groupby('CONVENIO')['DS_GUIA_PATH'].first().reset_index()
df_exemplos_aviso.rename(columns={'DS_GUIA_PATH': 'LINK_EXEMPLO'}, inplace=True)

# 4. Une a tabela de contagem com a de exemplos
df_final_aviso = pd.merge(df_contagem_aviso, df_exemplos_aviso, on='CONVENIO')

# 5. Aplica a formatação para deixar o link clicável e o percentual com o símbolo '%'
# (Reutilizando a função 'tornar_clicavel' que definimos antes)
tabela_formatada_aviso = df_final_aviso.style.format({
    'LINK_EXEMPLO': tornar_clicavel,
    'PERCENTUAL (%)': '{:.2f}%'.format
})

print("--- Tabela Detalhada de Convênios (Apenas com CD_AVISO > 0) ---")
display(tabela_formatada_aviso)
```

```python
import os
import json
import boto3
from dotenv import load_dotenv
import requests
import base64
import io # Necessário para manipulação de bytes em memória
from pdf2image import convert_from_bytes # A nova importação

# Carrega as variáveis do arquivo .env
load_dotenv()

# Lê as variáveis do ambiente
aws_access_key_id = os.getenv('AWS_ACCESS_KEY_ID')
aws_secret_access_key = os.getenv('AWS_SECRET_ACCESS_KEY')
aws_region = os.getenv('AWS_BEDROCK_REGION', 'us-east-1') # Valor padrão: us-east-1
model_id = os.getenv('BEDROCK_MODEL_ID')

# Validação para garantir que as variáveis foram carregadas
if not all([aws_access_key_id, aws_secret_access_key, model_id]):
    print("❌ Erro: Certifique-se de que o arquivo .env está criado e configurado corretamente.")
else:
    print("✅ Variáveis de ambiente carregadas com sucesso.")

try:
    bedrock_runtime = boto3.client(
        service_name='bedrock-runtime', 
        aws_access_key_id=aws_access_key_id,
        aws_secret_access_key=aws_secret_access_key,
        region_name=aws_region
    )
    print(f"✅ Cliente AWS Bedrock inicializado com sucesso na região '{aws_region}'.")
except Exception as e:
    print(f"❌ Erro ao inicializar o cliente Bedrock: {e}")


def carregar_imagem_da_url(url):
    """
    Baixa um arquivo de uma URL. Se for PDF, converte a primeira página para imagem (PNG).
    Se já for imagem, usa diretamente. Retorna os bytes e o formato de mídia.
    """
    try:
        response = requests.get(url, timeout=15)
        response.raise_for_status()
        
        file_bytes = response.content
        content_type = response.headers.get('content-type', '')

        # Verifica se o conteúdo é um PDF
        if 'application/pdf' in content_type or url.lower().endswith('.pdf'):
            print("📄 PDF detectado. Convertendo a primeira página para imagem PNG...")
            
            # Converte os bytes do PDF para uma lista de imagens (formato PIL)
            images = convert_from_bytes(file_bytes, first_page=1, last_page=1)
            
            if images:
                image = images[0]
                # Salva a imagem em um buffer de bytes em memória, no formato PNG
                buffer = io.BytesIO()
                image.save(buffer, format='PNG')
                image_bytes = buffer.getvalue()
                
                print("✅ Conversão de PDF para PNG concluída.")
                # O novo media type é sempre PNG após a conversão
                return image_bytes, 'image/png'
            else:
                print("❌ Não foi possível extrair imagens do PDF.")
                return None, None

        # Se for um tipo de imagem já aceito, usa diretamente
        elif content_type in ['image/jpeg', 'image/png', 'image/gif', 'image/webp']:
            print(f"✅ Imagem ({content_type}) baixada com sucesso de {url}")
            return file_bytes, content_type
        
        else:
            print(f"⚠️ Tipo de arquivo não suportado: {content_type}")
            return None, None

    except requests.exceptions.RequestException as e:
        print(f"❌ Erro ao baixar o arquivo: {e}")
        return None, None
    except Exception as e:
        print(f"❌ Erro no processamento do arquivo (verifique a instalação do Poppler): {e}")
        return None, None

print("Funções de apoio atualizadas para lidar com PDFs.")
```

```python
def extrair_dados_da_guia(image_bytes, media_type):
    """
    Envia a imagem de uma guia médica para a LLM multimodal no Bedrock
    e pede para extrair informações específicas em formato JSON.
    """
    # ... (o início da função continua igual) ...

    if not image_bytes:
        return {"erro": "Bytes da imagem não fornecidos."}

    # Codifica a imagem para Base64
    base64_image = base64.b64encode(image_bytes).decode('utf-8')

    # Prompt detalhado para guiar a LLM
    prompt_texto = """
    Você é um assistente especialista em analisar guias de autorização de convênios médicos (modelo TISS).
    Sua tarefa é extrair as seguintes informações da imagem fornecida e retornar **apenas e tão somente um objeto JSON válido**.

    Extraia os seguintes campos:
    - "procedimento_autorizado": O nome completo do procedimento principal autorizado (campo 25).
    - "paciente": O nome completo do paciente (campo 10).
    - "codigo_autorizado": Uma string contendo o código do procedimento e a quantidade (Um numero de 8 digitos)
    - "senha": O número da senha de autorização.
    - "validade_senha": A data de validade da senha (campo 6).
    - "observacoes_opme": Se houver alguma observação sobre OPME, transcreva-a. Se não houver menção, retorne "Pedido sem OPME".
    - "observacoes_gerais": O texto do campo "Indicação Clínica" (campo 21).
    - "usuario_finalizador": O nome do profissional solicitante (campo 14). Se não estiver presente, retorne null.

    O JSON de saída deve ter exatamente essa estrutura. Não inclua nenhuma outra explicação ou texto antes ou depois do JSON.
    """

    # Corpo da requisição para o modelo Claude 3 (multimodal)
    body = json.dumps({
        "anthropic_version": "bedrock-2023-05-31",
        "max_tokens": 2048,
        "messages": [
            {
                "role": "user",
                "content": [
                    {
                        "type": "image",
                        "source": {
                            "type": "base64",
                            "media_type": media_type,
                            "data": base64_image
                        }
                    },
                    {
                        "type": "text",
                        "text": prompt_texto
                    }
                ]
            }
        ]
    })

    try:
        # Invoca o modelo Bedrock
        response = bedrock_runtime.invoke_model(body=body, modelId=model_id)
        response_body = json.loads(response.get('body').read())
        
        # Extrai o texto da resposta
        resposta_texto_bruto = response_body['content'][0]['text']
        
        # --- LÓGICA DE LIMPEZA ADICIONADA AQUI ---
        # Procura o início do JSON (primeiro '{') e o fim (último '}')
        inicio_json = resposta_texto_bruto.find('{')
        fim_json = resposta_texto_bruto.rfind('}') + 1
        
        if inicio_json != -1 and fim_json != 0:
            json_limpo_str = resposta_texto_bruto[inicio_json:fim_json]
            dados_extraidos = json.loads(json_limpo_str)
            return dados_extraidos
        else:
            # Se não encontrar um JSON, lança um erro para ser capturado abaixo
            raise ValueError("Nenhum objeto JSON válido encontrado na resposta da LLM.")

    except Exception as e:
        resposta_bruta_para_erro = ""
        if 'response_body' in locals():
            resposta_bruta_para_erro = response_body.get('content', [{}])[0].get('text', '')
        
        print(f"❌ Ocorreu um erro ao chamar a LLM ou processar a resposta: {e}")
        return {"erro": str(e), "resposta_bruta": resposta_bruta_para_erro}

print("Função principal de extração atualizada com limpeza de JSON.")
```

```python
from PIL import Image
import io

def redimensionar_imagem(image_bytes, fator_zoom=3.0, limite_mb=5):
    """
    Redimensiona uma imagem por um fator, mas respeita um limite de tamanho em MB.
    
    :param image_bytes: A imagem original em bytes.
    :param fator_zoom: Fator para aumentar a imagem (ex: 2.0 para 200%).
    :param limite_mb: O tamanho máximo permitido para a imagem final em megabytes.
    :return: Os bytes da imagem redimensionada, ou os bytes da imagem original se o limite for excedido.
    """
    try:
        limite_bytes = limite_mb * 1024 * 1024
        imagem = Image.open(io.BytesIO(image_bytes))

        # Calcula as novas dimensões
        nova_largura = int(imagem.width * fator_zoom)
        nova_altura = int(imagem.height * fator_zoom)

        print(f"Redimensionando imagem de {imagem.size} para {(nova_largura, nova_altura)}...")
        
        # Redimensiona usando um filtro de alta qualidade (LANCZOS)
        imagem_redimensionada = imagem.resize((nova_largura, nova_altura), Image.Resampling.LANCZOS)

        # Salva a nova imagem em um buffer para verificar o tamanho
        buffer = io.BytesIO()
        imagem_redimensionada.save(buffer, format='PNG')
        bytes_redimensionados = buffer.getvalue()

        # Verifica se o tamanho final respeita o limite
        if len(bytes_redimensionados) > limite_bytes:
            print(f"⚠️ Aviso: A imagem redimensionada ({len(bytes_redimensionados) / 1024 / 1024:.2f} MB) excedeu o limite de {limite_mb} MB.")
            print("Enviando a imagem original para a LLM.")
            return image_bytes # Retorna a original se o limite for estourado
        else:
            print(f"✅ Imagem redimensionada com sucesso ({len(bytes_redimensionados) / 1024 / 1024:.2f} MB).")
            return bytes_redimensionados

    except Exception as e:
        print(f"❌ Erro ao redimensionar a imagem: {e}")
        return image_bytes # Retorna a original em caso de erro
```
