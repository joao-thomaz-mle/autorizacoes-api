# Part 1

load the dataframe into memory and bring it to a valid input for unstructured

```python
# load into memory
import sqlalchemy
import pandas as pd

# Create a database connection
engine = sqlalchemy.create_engine("sqlite:///docs/gold_autorizacao_database.sqlite")
merged_df_autorizacao_final = pd.read_sql("SELECT * FROM autorizacao", engine)
```

```python
merged_df_autorizacao_final.head()
```

```python
import requests
import numpy as np
def fetch_pdf_bytes(url):
    try:
        response = requests.get(url, timeout=15)
        if response.status_code == 200 and 'application/pdf' in response.headers.get('content-type', ''):
            return response.content
        else:
            return np.nan
    except Exception as e:
        print(f"Error fetching {url}: {e}")
        return np.nan

# Apply to all links in DS_GUIA_PATH
merged_df_autorizacao_final['DS_PDF_BYTES'] = merged_df_autorizacao_final['DS_GUIA_PATH'].apply(fetch_pdf_bytes)
merged_df_autorizacao_final.head()
```

```python
# localize the value of DS_PROCEDIMENTO of df merged_df_autorizacao_final FOR CD_AVISO_CIRURGIA =
ds_contato = merged_df_autorizacao_final.loc[merged_df_autorizacao_final['CD_AVISO_CIRURGIA'] == 857469, 'DS_CONTATO'].values[0]
ds_contato
```

```python
ds_procedimento = merged_df_autorizacao_final.loc[merged_df_autorizacao_final['CD_AVISO_CIRURGIA'] == 857469, 'DS_PROCEDIMENTO'].values[0]
ds_procedimento
```

```python
# Create a new column with clickable links for DS_GUIA_PATH
def make_clickable(url):
    if pd.notna(url):
        return f'<a href="{url}" target="_blank">{url}</a>'
    return ""
merged_df_autorizacao_final['DS_GUIA_PATH_CLICKABLE'] = merged_df_autorizacao_final['DS_GUIA_PATH'].apply(make_clickable)
from IPython.display import display, HTML
display(HTML(merged_df_autorizacao_final[['DS_GUIA_PATH', 'DS_GUIA_PATH_CLICKABLE']].head(10).to_html(escape=False)))
```

# Part 2 - processing with unstructured

```python
import io
from unstructured.partition.pdf import partition_pdf

# Get the PDF bytes for the desired row
pdf_bytes = merged_df_autorizacao_final["DS_PDF_BYTES"][merged_df_autorizacao_final['CD_AVISO_CIRURGIA'] == 857469].values[0]

# Wrap the bytes in a BytesIO object
pdf_file_like = io.BytesIO(pdf_bytes)

# Now pass to partition_pdf
elements = partition_pdf(file=pdf_file_like, languages=["portuguese"])

# Join the text
full_text = "\n".join([el.text for el in elements if el.text])
```

```python
print(full_text)
```

```python
# taking this output and processing trough a llm

import sys

# add app to path so we dont get the error ModuleNotFoundError: No module named 'app'
from pathlib import Path
sys.path.append(str(Path.cwd().parent))
from app.utils.llm_service import AnthropicLLMService
from app.utils.aws_services_handler import create_boto3_client
from app.utils.config import load_config, AppConstants
from app.utils.logger import get_logger
from app.utils.system_prompts.autorizacao_prompt import Prompts


logger = get_logger(name=__name__)
config_vars = load_config()
```

```python
app_constants = AppConstants()
bedrock_client = create_boto3_client("bedrock-runtime", config_vars)
autorizacao_prompt = Prompts.autorizacao_extraction_prompt

llm_instance = AnthropicLLMService(
    model_id=app_constants.BEDROCK_DEFAULT_MODEL_ID,
    model_version=app_constants.BEDROCK_DEFAULT_MODEL_VERSION,
    client=bedrock_client,
    system_prompt=autorizacao_prompt,
    max_tokens=app_constants.MAX_TOKENS,
    temperature=app_constants.TEMPERATURE,
    budget_tokens=app_constants.BUDGET_TOKENS
)
```

```python
# testing a single answer of the llm

results = llm_instance.invoke_model(
    input_str=full_text
)
```
