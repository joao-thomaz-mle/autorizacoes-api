```python
import requests
import numpy as np
def fetch_pdf_bytes(url):
    try:
        response = requests.get(url, timeout=15)
        if response.status_code == 200 and 'application/pdf' in response.headers.get('content-type', ''):
            return response.content
        else:
            return np.nan
    except Exception as e:
        print(f"Error fetching {url}: {e}")
        return np.nan

# Apply to all links in DS_GUIA_PATH
merged_df_autorizacao_final['DS_PDF_BYTES'] = merged_df_autorizacao_final['DS_GUIA_PATH'].apply(fetch_pdf_bytes)
merged_df_autorizacao_final.head()
```

```python
# Extract ground truth fields from DS_CONTATO
import re

def extract_ds_contato_fields(ds_contato_text):
    """Extract structured fields from DS_CONTATO text using regex patterns"""
    if pd.isna(ds_contato_text):
        return {field + '_gt': None for field in ['paciente', 'codigo_autorizado', 'senha', 'validade_senha', 'observacoes_opme', 'observacoes_gerais', 'usuario_finalizador']}
    
    extracted = {}
    
    # Extract Paciente
    paciente_match = re.search(r'Paciente:\s*(.+?)(?=\n|CÓDIGO)', ds_contato_text, re.IGNORECASE)
    extracted['paciente_gt'] = paciente_match.group(1).strip() if paciente_match else None
    
    # Extract CÓDIGO AUTORIZADO
    codigo_match = re.search(r'CÓDIGO AUTORIZADO:\s*(.+?)(?=\n\n|SENHA)', ds_contato_text, re.DOTALL | re.IGNORECASE)
    extracted['codigo_autorizado_gt'] = codigo_match.group(1).strip() if codigo_match else None
    
    # Extract SENHA
    senha_match = re.search(r'SENHA:\s*(.+?)(?=\n|VALIDADE)', ds_contato_text, re.IGNORECASE)
    extracted['senha_gt'] = senha_match.group(1).strip() if senha_match else None
    
    # Extract VALIDADE DA SENHA
    validade_match = re.search(r'VALIDADE DA SENHA:\s*(.+?)(?=\n|OBSERVAÇÕES)', ds_contato_text, re.IGNORECASE)
    extracted['validade_senha_gt'] = validade_match.group(1).strip() if validade_match else None
    
    # Extract OBSERVAÇÕES OPME
    opme_match = re.search(r'OBSERVAÇÕES OPME:\s*(.+?)(?=\n|OBSERVAÇÕES GERAIS)', ds_contato_text, re.IGNORECASE)
    extracted['observacoes_opme_gt'] = opme_match.group(1).strip() if opme_match else None
    
    # Extract OBSERVAÇÕES GERAIS
    gerais_match = re.search(r'OBSERVAÇÕES GERAIS:\s*(.+?)(?=\n|NOME USUÁRIO)', ds_contato_text, re.IGNORECASE)
    extracted['observacoes_gerais_gt'] = gerais_match.group(1).strip() if gerais_match else None
    
    # Extract NOME USUÁRIO FINALIZOU O PEDIDO
    usuario_match = re.search(r'NOME USUÁRIO FINALIZOU O PEDIDO:\s*(.+?)(?=\n|$)', ds_contato_text, re.IGNORECASE)
    extracted['usuario_finalizador_gt'] = usuario_match.group(1).strip() if usuario_match else None
    
    return extracted

# Apply extraction to all rows in eval_df
gt_columns = ['paciente_gt', 'codigo_autorizado_gt', 'senha_gt', 'validade_senha_gt', 'observacoes_opme_gt', 'observacoes_gerais_gt', 'usuario_finalizador_gt']

# Initialize ground truth columns
for col in gt_columns:
    merged_df_autorizacao_final[col] = None

# Extract fields for each row
for idx, row in merged_df_autorizacao_final.iterrows():
    extracted_fields = extract_ds_contato_fields(row['DS_CONTATO'])
    for field_name, value in extracted_fields.items():
        merged_df_autorizacao_final.at[idx, field_name] = value

print("Ground truth extraction completed!")
print("\nSample of extracted fields:")
merged_df_autorizacao_final[['CD_AVISO_CIRURGIA'] + gt_columns].head()
```

```python
# Step 1: Create subset of first 5 examples
eval_df = merged_df_autorizacao_final[:5].copy()
print(f"Created evaluation dataset with {len(eval_df)} rows")

# Initialize new columns for LLM results
llm_columns = [
    'procedimento_autorizado_llm',
    'paciente_llm', 
    'codigo_autorizado_llm',
    'senha_llm',
    'validade_senha_llm',
    'observacoes_opme_llm',
    'observacoes_gerais_llm',
    'usuario_finalizador_llm'
]

for col in llm_columns:
    eval_df[col] = None

print("Initialized LLM result columns")
eval_df.head()
```

```python
# Step 2 & 3: Process each row through the pipeline
import re
from tqdm import tqdm

def process_single_row(row_index, pdf_bytes):
    """Process a single row through the PDF->Text->LLM pipeline"""
    try:
        # Convert PDF bytes to text using unstructured
        pdf_file_like = io.BytesIO(pdf_bytes)
        elements = partition_pdf(file=pdf_file_like, languages=["portuguese"])
        full_text = "\n".join([el.text for el in elements if el.text])
        
        # Send to LLM
        results = llm_instance.invoke_model(input_str=full_text)
        
        # Clean special characters from specific fields
        if 'codigo_autorizado' in results:
            results["codigo_autorizado"] = re.sub(r'[^a-zA-Z0-9]', '', results["codigo_autorizado"])
        if 'senha' in results:
            results["senha"] = re.sub(r'[^a-zA-Z0-9]', '', results["senha"])
            
        return results
        
    except Exception as e:
        print(f"Error processing row {row_index}: {e}")
        return {col.replace('_llm', ''): None for col in llm_columns}

# Process each row
for idx, row in tqdm(eval_df.iterrows(), total=len(eval_df), desc="Processing PDFs"):
    pdf_bytes = row['DS_PDF_BYTES']
    
    # Skip if no PDF bytes available
    if pd.isna(pdf_bytes):
        print(f"Skipping row {idx}: No PDF bytes available")
        continue
        
    # Process the row
    llm_results = process_single_row(idx, pdf_bytes)
    
    # Add results to dataframe
    for field_name, value in llm_results.items():
        col_name = f"{field_name}_llm"
        if col_name in eval_df.columns:
            eval_df.at[idx, col_name] = value
    
    print(f"Completed row {idx} - CD_AVISO_CIRURGIA: {row['CD_AVISO_CIRURGIA']}")

print("Pipeline processing completed!")
eval_df[['CD_AVISO_CIRURGIA'] + llm_columns].head()
```

# Part 3 - processing with textract pdf extractor

```python

# load
import sqlalchemy
import pandas as pd
# Create a database connection to the correct database
engine = sqlalchemy.create_engine("sqlite:///merged_autorizacao_database.sqlite")


merged_df_autorizacao_final = pd.read_sql("SELECT * FROM autorizacao", engine)
```

```python
# trying to use the mudular code to get the workflow feeling
import sys

# add app to path so we dont get the error ModuleNotFoundError: No module named 'app'
from pathlib import Path
sys.path.append(str(Path.cwd().parent))
from app.utils.process_images import process_blobs
from app.utils.textract_service import TextractPDFAnalyzer
from app.utils.llm_service import AnthropicLLMService
from app.utils.aws_services_handler import create_boto3_client
from app.utils.config import load_config, AppConstants
from app.utils.logger import get_logger
from app.utils.system_prompts.autorizacao_prompt import Prompts


logger = get_logger(name=__name__)
config_vars = load_config()
```

```python
app_constants = AppConstants()
textract_client = create_boto3_client("textract", config_vars)
bedrock_client = create_boto3_client("bedrock-runtime", config_vars)
s3_client = create_boto3_client("s3", config_vars)
bucket_name = "autorizacoes"

# Initialize the TextractPDFAnalyzer
pdf_analyzer = TextractPDFAnalyzer(
    textract_client=textract_client,
    s3_client=s3_client,
    bucket_name="autorizacoes-textract",
    bucket_folder="guias-pdf"
)

autorizacao_prompt = Prompts.autorizacao_extraction_prompt


llm_instance = AnthropicLLMService(
    model_id=app_constants.BEDROCK_DEFAULT_MODEL_ID,
    model_version=app_constants.BEDROCK_DEFAULT_MODEL_VERSION,
    client=bedrock_client,
    system_prompt=autorizacao_prompt,
    max_tokens=app_constants.MAX_TOKENS,
    temperature=app_constants.TEMPERATURE,
    budget_tokens=app_constants.BUDGET_TOKENS
)
```

```python
textract_results = {}
llm_results = {}

# Process first 3 rows from the dataframe
for index, row in merged_df_autorizacao_final.iterrows():
    pdf_bytes = row['DS_PDF_BYTES']
    file_id = str(row['CD_AVISO_CIRURGIA'])
    
    if pd.isna(pdf_bytes):
        print(f"Skipping row {index}: No PDF bytes available")
        continue
    
    # Extract forms data using Textract
    forms_data = pdf_analyzer.extract_forms_data_from_pdf(
        pdf_bytes=pdf_bytes,
        file_id=file_id
    )
    
    textract_results[file_id] = forms_data
    
    # Process with LLM
    forms_data_str = str(forms_data)
    llm_response = llm_instance.invoke_model(input_str=forms_data_str)
    llm_results[file_id] = llm_response

# Save to dataframes
textract_results_df = pd.DataFrame.from_dict(textract_results, orient="index")
textract_results_df.index.name = "CD_AVISO_CIRURGIA"
textract_results_df = textract_results_df.reset_index()

llm_results_df = pd.DataFrame.from_dict(llm_results, orient="index")
llm_results_df.index.name = "CD_AVISO_CIRURGIA"
llm_results_df = llm_results_df.reset_index()

# Expand LLM results into separate columns
llm_expanded_df = pd.json_normalize(llm_results_df.iloc[:, 1:].to_dict('records'))
llm_expanded_df['CD_AVISO_CIRURGIA'] = llm_results_df['CD_AVISO_CIRURGIA']

# Add suffix to distinguish LLM columns
llm_expanded_df = llm_expanded_df.add_suffix('_llm').rename(columns={'CD_AVISO_CIRURGIA_llm': 'CD_AVISO_CIRURGIA'})

# Convert CD_AVISO_CIRURGIA to int to match the original dataframe type
llm_expanded_df['CD_AVISO_CIRURGIA'] = llm_expanded_df['CD_AVISO_CIRURGIA'].astype(int)

# Inner join with original dataframe
textract_eval_df = merged_df_autorizacao_final.merge(
    llm_expanded_df, 
    on='CD_AVISO_CIRURGIA', 
    how='inner'
)

print(f"Processing completed! Created textract_eval_df with {len(textract_eval_df)} rows")
print(f"LLM columns added: {[col for col in textract_eval_df.columns if col.endswith('_llm')]}")
textract_eval_df.head()
```

```python
# able to see all pandas cols
pd.set_option('display.max_columns', None)
# apply cleaning on the value from cols senha_llm using re to clean white spaces and special characters
import re

def clean_text(text):
    if pd.notna(text):
        text = re.sub(r'\s+', ' ', text)  # Replace multiple spaces with a single space
        text = re.sub(r'[^\w\s]', '', text)  # Remove special characters
        return text.strip()
    return ""

textract_eval_df['senha_llm'] = textract_eval_df['senha_llm'].apply(clean_text)
textract_eval_df.head()
```

```python
#save textract_eval_df to sql
import sqlalchemy
eval_engine = sqlalchemy.create_engine("sqlite:///textract_eval_database.sqlite")
textract_eval_df.to_sql("textract_eval", con=eval_engine, if_exists="replace", index=False)
```
